```python
from src.data_loader import load_price_volume, find_extreme_moves
from src.features import build_all_factors
from src.ic_analysis import forward_return, calc_ic, ic_summary_table, align_common_obs, multiple_testing_correction

# 1. Load dữ liệu — universe mặc định giờ là 30 mã VN30
price_df, open_df, volume_df = load_price_volume()

# 2. Data audit — dùng số này để viết lại reports/data_audit.md
extreme = find_extreme_moves(price_df)
print(f"Số phiên biến động > 7%: {len(extreme)}")
print(extreme.head(10))

# 3. Build 12 factor
factors = build_all_factors(price_df, open_df, volume_df)

# 4. IC — dùng số này để viết lại bảng "Kết quả chính" trong README
fwd_ret = forward_return(price_df, n_fwd=5)
ic_series = {name: calc_ic(f, fwd_ret) for name, f in factors.items()}

summary_raw = ic_summary_table(ic_series)          # riêng lẻ, N Obs có thể lệch nhau
ic_common = align_common_obs(ic_series)
summary_common = ic_summary_table(ic_common)        # bảng dùng để so sánh công bằng — lấy số này cho README
correction = multiple_testing_correction(summary_common)

print(summary_common)
print(correction)
```
